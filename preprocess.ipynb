```python
import pandas as pd 
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', 50) 
df=pd.read_csv('apartments_raw_data.csv')
```

```python
df.info()
```

```python
df.fillna({'loggia_area_m2':0}, inplace=True)
df.fillna({'cellar_area_m2':0}, inplace=True)
df.fillna({'total_floors':df['total_floors'].mode()[0] if not df['total_floors'].mode().empty else 0}, inplace=True)
df.fillna({'floor_number':df['floor_number'].mode()[0] if not df['floor_number'].mode().empty else 2}, inplace=True)
df.fillna({"total_area_m2": df['usable_area_m2']}, inplace=True)
df.fillna({'is_furnished':0}, inplace=True)
df.fillna({'energy_class':"unknown"}, inplace=True)
```

```python
df.drop(columns=['construction_year'], inplace=True)
df.drop(columns=['currency'], inplace=True)
df.drop(columns=['is_topped'], inplace=True)
df.drop(columns=['updated'], inplace=True)
df.drop(columns=['created_at'], inplace=True)
df.drop(columns=['energy_efficiency_rating'], inplace=True)
df.drop(columns=['municipality_id'], inplace=True)
df.drop(columns=['id'], inplace=True)
df.drop(columns=['web_link'], inplace=True)
df = df[df['total_area_m2'] <= 500].copy()

```

```python
df.info()
```

```python
df_clean = df.copy()
df_clean['category_sub'] = df_clean['category_sub'].replace(['1+kk'], '1')
df_clean['category_sub'] = df_clean['category_sub'].replace(['1+1', '2+kk'], '2')
df_clean['category_sub'] = df_clean['category_sub'].replace(['2+1', '3+kk'], '3')
df_clean['category_sub'] = df_clean['category_sub'].replace(['3+1', '4+kk'], '4')
df_clean['category_sub'] = df_clean['category_sub'].replace(['4+1', '5+kk','6-a-vice','atypicky',"5+1"], '5+')
```
